# 01 - Data inspection
Load the source CSVs, inspect their structure, and record data-quality findings before analysis.

```python
from pathlib import Path
import pandas as pd

RAW_DIR = Path('../data/raw')
csv_files = sorted(RAW_DIR.glob('*.csv'))
print(f'Found {len(csv_files)} CSV files')
[p.name for p in csv_files]
```

```python
tables = {}
for path in csv_files:
    df = pd.read_csv(path)
    tables[path.stem] = df
    print(f'\\n{path.name}: {df.shape[0]:,} rows x {df.shape[1]} columns')
    display(df.head())
```

```python
# Build a compact profiling summary for each table.
for name, df in tables.items():
    print(f'\\n## {name}')
    display(pd.DataFrame({
        'dtype': df.dtypes.astype(str),
        'missing': df.isna().sum(),
        'missing_pct': (df.isna().mean() * 100).round(2),
        'distinct': df.nunique(dropna=True)
    }))
    print('Duplicate rows:', int(df.duplicated().sum()))
```
